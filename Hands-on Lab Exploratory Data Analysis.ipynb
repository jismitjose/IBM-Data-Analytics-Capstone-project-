#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
# Step 3: Identify missing values in the critical columns
critical_cols = ["Employment", "JobSat", "RemoteWork"]
print(df[critical_cols].isnull().sum())
print(df[critical_cols].dtypes)
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
# Step 4: Convert YearsCodePro to numeric and summarize it
years = df["YearsCodePro"].replace({
    "Less than 1 year": 0.5,
    "More than 50 years": 51
})
df["YearsCodePro_num"] = pd.to_numeric(years, errors="coerce")
print(df["YearsCodePro_num"].describe())
# Create experience ranges and calculate the median JobSat for each
bins = [0, 5, 10, 20, float("inf")]
labels = ["0-5", "5-10", "10-20", ">20"]
df["ExperienceRange"] = pd.cut(df["YearsCodePro_num"], bins=bins, labels=labels, include_lowest=True)

median_jobsat = df.groupby("ExperienceRange", observed=True)["JobSat"].median()
print(median_jobsat)
print(df["ExperienceRange"].value_counts())
plt.figure(figsize=(8, 5))
median_jobsat.plot(kind="bar", color="#0072B2")
plt.title("Median Job Satisfaction by Years of Professional Coding Experience")
plt.xlabel("Years of professional coding experience")
plt.ylabel("Median JobSat")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
# Step 5: Count plot of JobSat
import seaborn as sns
import matplotlib.pyplot as plt

jobsat_values = df["JobSat"].dropna()
order = sorted(jobsat_values.unique())

plt.figure(figsize=(9, 5))
sns.countplot(x=jobsat_values.astype(int), order=[int(v) for v in order], color="#0072B2")
plt.title("Distribution of Job Satisfaction (JobSat)")
plt.xlabel("Job satisfaction rating (0 = lowest, 10 = highest)")
plt.ylabel("Number of respondents")
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
## Write your code here
# Step 6: Count plot of remote work distribution
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 4))
sns.countplot(y="RemoteWork", data=df, order=df["RemoteWork"].value_counts().index, color="#0072B2")
plt.title("Remote Work Distribution")
plt.xlabel("Number of respondents")
plt.ylabel("RemoteWork")
plt.tight_layout()
plt.show()
# Cross-tabulate RemoteWork by Employment (top 5 employment types)
top_employment = df["Employment"].value_counts().head(5).index
subset = df[df["Employment"].isin(top_employment)]

crosstab_counts = pd.crosstab(subset["Employment"], subset["RemoteWork"])
print(crosstab_counts)
# Row percentages: share of each employment type by remote work preference
crosstab_pct = pd.crosstab(subset["Employment"], subset["RemoteWork"], normalize="index") * 100
print(crosstab_pct.round(1))

crosstab_pct.plot(kind="barh", stacked=True, figsize=(10, 5), color=["#0072B2", "#E69F00", "#009E73"])
plt.title("Remote Work Preference by Employment Type (%)")
plt.xlabel("Percent of respondents")
plt.ylabel("Employment")
plt.legend(title="RemoteWork", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
# Step 7: Filter by country and split the languages
top_countries = df["Country"].value_counts().head(5).index
print("Top 5 countries:", list(top_countries))

sub = df[df["Country"].isin(top_countries)][["Country", "LanguageHaveWorkedWith"]].dropna()
sub["Language"] = sub["LanguageHaveWorkedWith"].str.split(";")
exploded = sub.explode("Language")
print("Respondents with language data:", sub.shape[0])
# Top 5 languages in each country
for country in top_countries:
    top5 = exploded[exploded["Country"] == country]["Language"].value_counts().head(5)
    print(country)
    print(top5)
    print()
# Heatmap: % of respondents in each country who have worked with each language

counts = exploded.groupby(["Country", "Language"]).size().unstack(fill_value=0)
respondents = sub["Country"].value_counts()
pct = counts.div(respondents, axis=0) * 100

top_langs = exploded["Language"].value_counts().head(10).index

plt.figure(figsize=(11, 5))
sns.heatmap(pct[top_langs], annot=True, fmt=".0f", cmap="Blues", cbar_kws={"label": "% of respondents"})
plt.title("Top Programming Languages by Country (% of respondents)")
plt.xlabel("Language")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
# Step 8: Correlation between YearsCodePro and JobSatPoints_1

years = df["YearsCodePro"].replace({"Less than 1 year": 0.5, "More than 50 years": 51})
df["YearsCodePro_num"] = pd.to_numeric(years, errors="coerce")

corr_data = df[["YearsCodePro_num", "JobSatPoints_1"]].dropna()
print("Rows used:", corr_data.shape[0])
print("Correlation:", corr_data["YearsCodePro_num"].corr(corr_data["JobSatPoints_1"]))
# Scatter plot
plt.figure(figsize=(9, 5))
plt.scatter(corr_data["YearsCodePro_num"], corr_data["JobSatPoints_1"], alpha=0.15, color="#0072B2")
plt.title("Years of Professional Coding Experience vs JobSatPoints_1")
plt.xlabel("Years of professional coding experience")
plt.ylabel("JobSatPoints_1")
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
# Step 9: Short labels for EdLevel (original column unchanged)
def short_edlevel(value):
    if pd.isnull(value):
        return value
    for prefix, label in [("Primary", "Primary"), ("Secondary", "Secondary"),
                          ("Some college", "Some College"), ("Associate", "Associate"),
                          ("Bachelor", "Bachelor"), ("Master", "Master"),
                          ("Professional", "Professional/Doctorate")]:
        if value.startswith(prefix):
            return label
    return "Other"

df["EdLevel_short"] = df["EdLevel"].apply(short_edlevel)
print(df["EdLevel_short"].value_counts())
# Cross-tabulation: EdLevel vs top 5 Employment types
top_employment = df["Employment"].value_counts().head(5).index
subset = df[df["Employment"].isin(top_employment)]

crosstab_counts = pd.crosstab(subset["EdLevel_short"], subset["Employment"])
print(crosstab_counts)
# Row percentages: share of each education level by employment type, shown as a heatmap

edu_order = ["Primary", "Secondary", "Some College", "Associate", "Bachelor",
             "Master", "Professional/Doctorate", "Other"]
crosstab_pct = pd.crosstab(subset["EdLevel_short"], subset["Employment"], normalize="index") * 100
crosstab_pct = crosstab_pct.reindex(edu_order)

plt.figure(figsize=(12, 5))
sns.heatmap(crosstab_pct, annot=True, fmt=".0f", cmap="Blues", cbar_kws={"label": "% within education level"})
plt.title("Employment Type by Education Level (% within each education level)")
plt.xlabel("Employment")
plt.ylabel("Education level")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```
