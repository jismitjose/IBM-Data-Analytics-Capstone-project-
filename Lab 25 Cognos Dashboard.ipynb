```python
import glob, os
import pandas as pd

matches = glob.glob("survey_data_updated*.csv")
if not matches:
    raise FileNotFoundError("Upload your survey_data_updated CSV to the notebook's file browser first.")
clean = pd.read_csv(matches[0])
print("Source:", matches[0], "| rows:", len(clean), "(expected 18846)")

cols = {c.lower(): c for c in clean.columns}
fields = ["LanguageHaveWorkedWith", "DatabaseHaveWorkedWith", "PlatformHaveWorkedWith",
          "WebframeHaveWorkedWith", "LanguageWantToWorkWith", "DatabaseWantToWorkWith",
          "PlatformWantToWorkWith", "WebframeWantToWorkWith"]

missing = [f for f in fields if f.lower() not in cols]
print("Missing columns:", missing if missing else "none")

parts = []
for f in fields:
    if f.lower() not in cols:
        continue
    s = (clean[cols[f.lower()]].dropna()
                .str.split(";")
                .explode()
                .str.strip()
                .reset_index(drop=True))
    s = s[s != ""]
    t = s.value_counts().rename_axis("Technology").reset_index(name="Respondents")
    t.insert(0, "Category", f)
    parts.append(t)

out = pd.concat(parts, ignore_index=True)
out.to_csv("tech-counts.csv", index=False)
print("Saved tech-counts.csv:", out.shape)

for f in fields:
    print("\nTop 10:", f)
    print(out[out["Category"] == f].head(10).to_string(index=False))
```
