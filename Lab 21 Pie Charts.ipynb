### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
db_counts = (df["DatabaseWantToWorkWith"].dropna().str.split(";").explode().str.strip().value_counts())
top5 = db_counts.head(5)
print(top5)

plt.figure(figsize=(7, 7))
plt.pie(top5.values, labels=top5.index, autopct="%1.1f%%", startangle=90)
plt.title("Top 5 Databases Respondents Want to Work With")
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
dev_counts = (df["DevType"].dropna().str.split(";").explode().str.strip().value_counts())
top5_dev = dev_counts.head(5)
print(top5_dev)

plt.figure(figsize=(8, 8))
plt.pie(top5_dev.values, labels=top5_dev.index, autopct="%1.1f%%", startangle=90)
plt.title("Top 5 Developer Roles (DevType)")
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
##Write your code here
# Option1
os_counts = (df["OpSysProfessional use"].dropna().str.split(";").explode().str.strip().value_counts())
print(os_counts)

plt.figure(figsize=(8, 8))
plt.pie(os_counts.values, labels=os_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Operating Systems Used for Professional Use")
plt.show()
# Option 2 
os_counts = (df["OpSysProfessional use"].dropna().str.split(";").explode().str.strip().value_counts())
print(os_counts)

top8 = os_counts.head(8)
other = os_counts.iloc[8:].sum()
plot_data = pd.concat([top8, pd.Series({"All others": other})])

plt.figure(figsize=(8, 8))
plt.pie(plot_data.values, labels=plot_data.index, autopct="%1.1f%%", startangle=90)
plt.title("Operating Systems Used for Professional Use (top 8 + all others)")
plt.show()
```
