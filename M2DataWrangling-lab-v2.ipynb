### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
# 2.1 Summarize the dataset: data types, counts and missing values
summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Non-Null Count": df.count(),
    "Missing Values": df.isnull().sum()
})
print("Shape:", df.shape)
print(summary)
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
# 2.2 Basic statistics for numerical columns
print(df.describe())
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# 3.1 Identify inconsistent or irrelevant entries in Country
print("Unique countries:", df["Country"].nunique())
print("Missing Country:", df["Country"].isnull().sum())
# Leading/trailing spaces
country_clean = df["Country"].dropna()
print("Entries with extra spaces:", (country_clean != country_clean.str.strip()).sum())

# Least frequent values, where odd or inconsistent entries tend to show up
print(df["Country"].value_counts().tail(15))

# Long official names that are usually standardized
long_names = [c for c in df["Country"].dropna().unique() if len(c) > 25]
print(long_names)
# EdLevel: show the current values first
print(df["EdLevel"].value_counts())
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
# 3.2 Standardize Country
country_mapping = {
    "United States of America": "USA",
    "United Kingdom of Great Britain and Northern Ireland": "UK",
    "Russian Federation": "Russia",
    "Republic of Korea": "South Korea",
    "Democratic People's Republic of Korea": "North Korea",
    "Viet Nam": "Vietnam",
    "Hong Kong (S.A.R.)": "Hong Kong",
    "Iran, Islamic Republic of...": "Iran",
    "Venezuela, Bolivarian Republic of...": "Venezuela",
    "Micronesia, Federated States of...": "Micronesia",
    "Syrian Arab Republic": "Syria",
    "United Republic of Tanzania": "Tanzania",
    "The former Yugoslav Republic of Macedonia": "North Macedonia",
    "Republic of North Macedonia": "North Macedonia",
    "Democratic Republic of the Congo": "DR Congo",
    "Lao People's Democratic Republic": "Laos",
}

df["Country"] = df["Country"].str.strip().replace(country_mapping)
print(df["Country"].value_counts().head(10))
print([c for c in df["Country"].dropna().unique() if len(c) > 25])
print(df["EdLevel"].value_counts())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# 4.1 One-hot encode the Employment column
employment_encoded = pd.get_dummies(df["Employment"], prefix="Employment")
print("Number of encoded columns:", employment_encoded.shape[1])
print(employment_encoded.head())

# Add the encoded columns to the dataframe
df = pd.concat([df, employment_encoded], axis=1)
print("New shape:", df.shape)
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# 5.1 Columns with the highest number of missing values
missing_counts = df.isnull().sum().sort_values(ascending=False)
print(missing_counts.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# 5.2 Impute missing values in ConvertedCompYearly with the median
print("Missing before:", df["ConvertedCompYearly"].isnull().sum())
median_comp = df["ConvertedCompYearly"].median()
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(median_comp)
print("Median used:", median_comp)
print("Missing after:", df["ConvertedCompYearly"].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# 5.3 Impute missing values in RemoteWork with the most frequent value
print("Missing before:", df["RemoteWork"].isnull().sum())
most_frequent_remote = df["RemoteWork"].mode()[0]
df["RemoteWork"] = df["RemoteWork"].fillna(most_frequent_remote)
print("Most frequent value:", most_frequent_remote)
print("Missing after:", df["RemoteWork"].isnull().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
# 6.1 Min-Max scaling of ConvertedCompYearly
comp_min = df["ConvertedCompYearly"].min()
comp_max = df["ConvertedCompYearly"].max()
df["ConvertedCompYearly_MinMax"] = (df["ConvertedCompYearly"] - comp_min) / (comp_max - comp_min)
print(df["ConvertedCompYearly_MinMax"].describe())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
# 6.2 Log-transform ConvertedCompYearly to reduce skewness
import numpy as np

df["ConvertedCompYearly_Log"] = np.log1p(df["ConvertedCompYearly"])
print("Skewness before:", df["ConvertedCompYearly"].skew())
print("Skewness after:", df["ConvertedCompYearly_Log"].skew())
print(df["ConvertedCompYearly_Log"].describe())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
# Check YearsCodePro values
print(df["YearsCodePro"].dtype)
print(df["YearsCodePro"].isnull().sum())
print(df["YearsCodePro"].value_counts().head(10))
# 7.1 Create ExperienceLevel from YearsCodePro
years = df["YearsCodePro"].replace({
    "Less than 1 year": 0.5,
    "More than 50 years": 51
})
years = pd.to_numeric(years, errors="coerce")

def experience_level(y):
    if pd.isnull(y):
        return "Unknown"
    if y < 3:
        return "Beginner"
    elif y < 10:
        return "Intermediate"
    else:
        return "Advanced"

df["ExperienceLevel"] = years.apply(experience_level)
print(df["ExperienceLevel"].value_counts())
```
